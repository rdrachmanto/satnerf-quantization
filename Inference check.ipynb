```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load CSV
df = pd.read_csv("logs/single-sens_metrics_avg.csv")
df["scene"] = "JAX_412"

# Choose your scene
scene = "JAX_412"

# Filter data for this scene
df_scene = df[df["scene"] == scene]

variant_map = {
    "original_model": "1",
    "model_group_1_fp16.trt": "2",
    "model_group_2_fp16.trt": "3",
    "model_group_3_fp16.trt": "4",
    "model_group_4_fp16.trt": "5",
    "model_group_6_fp16.trt": "6",
    "model_group_7_fp16.trt": "7",
    "model_group_8_fp16.trt": "8"
}

df_scene["variant_code"] = df_scene["variant"].map(variant_map)
df_scene = df_scene.sort_values("variant_code")

# Create figure and subplots
fig, axes = plt.subplots(1, 3, figsize=(10, 3))

for ax in axes:
    ax.grid(True, zorder=0)
    ax.set_axisbelow(True)
    ax.tick_params(direction="in")

# Plot PSNR
axes[0].bar(df_scene["variant_code"].astype(str), df_scene["avg_psnr"], color="#901E3E", edgecolor="black")
axes[0].set_ylabel("PSNR (dB)")
axes[0].set_title(f"Average PSNR for {scene}")

# Plot MAE
axes[1].bar(df_scene["variant_code"].astype(str), df_scene["avg_mae"], color="#1F7D53", edgecolor="black")
axes[1].set_ylabel("MAE")
axes[1].set_title(f"Average MAE for {scene}")

# Plot SSIM
axes[2].bar(df_scene["variant_code"].astype(str), df_scene["avg_ssim"], color="#F6B17A", edgecolor="black")
axes[2].set_ylabel("SSIM")
axes[2].set_title(f"Average SSIM for {scene}")

variant_explanation = "\n".join(
    [f"{code} = {name}" for name, code in variant_map.items()]
)

plt.figtext(1, 0.4, variant_explanation,
            horizontalalignment='left',
            fontsize=10,
            bbox=dict(facecolor='white', alpha=1))

plt.tight_layout()
plt.show()
```
